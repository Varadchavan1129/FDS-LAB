```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the student performance dataset
df = pd.read_csv("StudentPerformance_Updated.csv")

# Matplotlib: compare average scores across subjects
subjects = ["Mathematics", "ScienceScore", "EnglishScore"]
mean_scores = df[subjects].mean()

plt.figure(figsize=(8, 5))
plt.bar(mean_scores.index, mean_scores.values, color=["#2878B5", "#F39B35", "#4A9D6E"])
plt.title("Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average score")
plt.ylim(0, 100)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

# Seaborn: relationship between study hours and average score
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="StudyHours", y="AverageScore", hue="Gender", s=80)
plt.title("Study Hours and Average Score")
plt.xlabel("Study hours")
plt.ylabel("Average score")
plt.tight_layout()
plt.show()

# Seaborn: correlations among numeric performance measures
numeric_columns = ["Mathematics", "ScienceScore", "EnglishScore", "StudyHours", "AverageScore"]
plt.figure(figsize=(8, 6))
sns.heatmap(
    df[numeric_columns].corr(),
    annot=True,
    cmap="vlag",
    center=0,
    fmt=".2f"
) git
plt.title("Correlation Between Study and Performance Measures")
plt.tight_layout()
plt.show()
```
