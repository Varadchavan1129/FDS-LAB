```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('Smart_Eco_Sales_Dataset.csv')
print(df)
df = pd.DataFrame(df)
df
```

```python
print("First 3 rows:")
display(df.head(3))

print("\nLast 3 rows:")
display(df.tail(3))
```

```python

print("\nInfo:")
df.info()

print("\nDescribe:")
display(df.describe())

print("\nShape (rows, columns):", df.shape)
```

```python
print("loc example (rows 0-2, specific columns):")
display(df.loc[0:2, ["Quantity", "UnitPrice", "Revenue"]])

print("\niloc example (first 3 rows, first 3 columns):")
display(df.iloc[0:3, 0:3])


print("\nBoolean indexing (City == 'Pune'):")
display(df[df["City"] == "Pune"])


```

```python
print("Type",type(df))
print("columns",list(df.columns))
print("Index:", df.index.tolist())
```

```python
print("Missing values per column:")
display(df.isnull().sum())

print("\nRows with any missing values:")
display(df[df.isnull().any(axis=1)])

print("\nDropping rows with missing values (dropna):")
display(df.dropna())

print("\nFilling missing Quantity with the column mean (fillna):")
df_filled = df.copy()
df_filled["Quantity"] = df_filled["Quantity"].fillna(df_filled["Quantity"].mean())
display(df_filled)

```
