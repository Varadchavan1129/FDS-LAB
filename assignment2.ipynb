```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("StudentPerformance.csv")
```

```python
print(df.head())
print(df.info())    
print(df.describe())
print(df.shape)
```

```python
print(df["MathScore"])
print(df[["Name", "MathScore", "ScienceScore"]])
print(df.loc[0:4, ["Name", "MathScore"]])
print(df.iloc[0:5, 0:4])
```

```python
high_score = df[df["MathScore"] > 80]

print(high_score)
gender_avg = df.groupby("Gender")["MathScore"].mean()

print(gender_avg)
```

```python
summary = df.groupby("Gender").agg(
    Average_Math=("MathScore","mean"),
    Highest_Math=("MathScore","max"),
    Total_Students=("Gender","count")
)

print(summary)
```

```python
pivot = pd.pivot_table(
    df,
    values="MathScore",
    index="Gender",
    columns="StudyHours",
    aggfunc="mean"
)

print(pivot)
```

```python
df["AverageScore"] = (
    df["MathScore"] +
    df["ScienceScore"] +
    df["EnglishScore"]
) / 3
```

```python
df["Grade"] = np.where(df["AverageScore"] >= 75,"Pass","Fail")
df.rename(columns={"MathScore":"Mathematics"}, inplace=True)
df["StudyHours"] = df["StudyHours"].astype(int)
top_students = df.sort_values(
    by="AverageScore",
    ascending=False
)

print(top_students.head())
df.to_csv("StudentPerformance_Updated.csv", index=False)
```
